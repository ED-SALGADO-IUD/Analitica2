# AE2 — Manejo de archivos CSV, JSON y TXT

Carga local, actualización de datos y generación de archivos de salida usando tres clases.

```python
from pathlib import Path
from manejador_csv import ManejadorCSV
from manejador_json import ManejadorJSON
from generador_resumen_txt import GeneradorResumenTXT

carpeta = Path("datos")
url_csv = "https://huggingface.co/datasets/a4n9i/iris-tabular/resolve/main/iris.csv"
url_json = "https://huggingface.co/datasets/logiover/naukri-job-scraper-sample-data/resolve/main/sample.json"
print("URL del CSV:", url_csv)
print("URL del JSON:", url_json)
```

## 1. Importar CSV y comprobar el registro original

```python
csv_obj = ManejadorCSV()
ruta_csv = carpeta / "iris_original.csv"
cantidad_csv = csv_obj.importar(ruta_csv)
print("Registros CSV:", cantidad_csv)
print("Antes:", csv_obj.registros[0])
```

## 2. Modificar y guardar el CSV

```python
campo_csv = "Species"
cambio_csv = csv_obj.modificar(0, campo_csv, "setosa-revisada")
salida_csv = carpeta / "iris_modificado.csv"
csv_obj.guardar(salida_csv)
verificacion_csv = ManejadorCSV()
verificacion_csv.importar(salida_csv)
assert verificacion_csv.registros[0][campo_csv] == "setosa-revisada"
print("Cambio:", cambio_csv)
print("Después:", verificacion_csv.registros[0])
print("Archivo:", salida_csv)
```

## 3. Importar JSON y mostrar el primer registro

```python
json_obj = ManejadorJSON()
ruta_json = carpeta / "empleos_original.json"
cantidad_json = json_obj.importar(ruta_json)
print("Registros JSON:", cantidad_json)
print("Antes:", json_obj.registros[0])
```

## 4. Modificar y guardar el JSON

```python
campo_json = "jobTitle"
valor_anterior_json = json_obj.registros[0][campo_json]
cambio_json = json_obj.modificar(0, campo_json, valor_anterior_json + " (revisado)")
salida_json = carpeta / "empleos_modificado.json"
json_obj.guardar(salida_json)
verificacion_json = ManejadorJSON()
verificacion_json.importar(salida_json)
assert verificacion_json.registros[0][campo_json].endswith(" (revisado)")
print("Cambio:", cambio_json)
print("Después:", verificacion_json.registros[0])
print("Archivo:", salida_json)
```

## 5. Generar y leer resumen TXT

```python
info_csv = {"url": url_csv, "origen": str(ruta_csv), "salida": str(salida_csv), "cantidad": cantidad_csv, "cambios": csv_obj.cambios}
info_json = {"url": url_json, "origen": str(ruta_json), "salida": str(salida_json), "cantidad": cantidad_json, "cambios": json_obj.cambios}
ruta_resumen = carpeta / "resumen.txt"
GeneradorResumenTXT().generar(ruta_resumen, info_csv, info_json)
with open(ruta_resumen, "r", encoding="utf-8") as archivo:
    print(archivo.read())
```
